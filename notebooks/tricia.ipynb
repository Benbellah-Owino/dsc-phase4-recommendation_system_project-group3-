### 1.  Dataset Overview

The ml-latest-small dataset contains the following files:

ratings.csv: User ratings for movies (core dataset)

movies.csv: Movie titles and genres

tags.csv: User-generated tags for movies

links.csv: External identifiers (IMDb and TMDB)

Among these, ratings.csv is the primary dataset used to build recommendation models.
